```python
# question is to plot a scatter plot and add a grid into it 

sales = [100, 200, 150, 300, 120]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

plt.scatter(months,sales) ## remember everyone x axis is horizontal , y axis is vertical 
plt.grid(True) # alright remmeber that grids should be true not numbers
plt.xlabel('monthly')
plt.ylabel('sales')
plt.title("monthly_sales")
plt.show()
```

```python
ages = [22, 25, 27, 23, 35, 40, 45, 28, 30, 22, 25, 38]

plt.hist(ages,bins=5)
plt.title("monthly_sales")
plt.xlabel("months")
plt.ylabel('sales')
plt.grid(True)
plt.show()
```

```python
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales_2023 = [100, 200, 150, 300, 120]
sales_2024 = [150, 180, 200, 280, 160]


x=np.arange(len(months))
width = 0.4
plt.bar(x-width/2 , sales_2023,months ,width, label = "2023")
plt.bar(x+width/2 , sales_2024,months ,width , label = '2024')
plt.xlabel("yearly_sales")
plt.ylabel("month_wise")
plt.show()
```

```python
import pandas as pd 
import seaborn as sns
df = sns.load_dataset('tips')

df.info()
df.isnull().sum()
df.describe(include='all')
```

## from above describe function i can understand that there are 3 categorical values , 4 integer + float combined , therefore the bill would be one of the primary keys for this and cohort wise analysis can be done 

```python
sns.barplot(data=df , x='sex', y='total_bill' , hue='time', color="purple")
plt.xlabel("Gender")
plt.ylabel("bills")
plt.title("gender_comparision")
plt.show()
```

## From above we can see that males pay more bills compared to females and next question would be who tips more ?


```python
sns.barplot(data=df , x='sex', y='tip' , hue='smoker', color="red")
plt.xlabel("Gender")
plt.ylabel("tips")
plt.title("gender_comparision")
plt.show()
```

```python
sns.countplot(data=df , x='day', hue='sex', color="red")
plt.xlabel('Day')
plt.ylabel('Gender')
plt.title("Day-Time-Comparision")
plt.show()
```

## As we can see users customers come a lot on thrusday's and Fridays and mostly females come on Thrusday followed by Satruday , where as men come around on saturday and Sunday so putting discounts on thursday/friday will drive more week day traffic

```python
sns.scatterplot(data=df , x='total_bill', y='tip', color="red")
plt.xlabel('bills')
plt.ylabel('tips')
plt.title("bill-tip-ratio")
plt.show()
```

### From the above image i am able to gather insights that smaller bills leads to smaller tips , higher bills lead to higher tips but the frequency in which we will get those will less than compared to frequency in which we will get for smaller bills and smaller tips , business suggestion would be to improve this specific aspect by focusing more on proper service and categoring towards a bit more higher bills faster ( assumption that fast + serivce could be a reason )
