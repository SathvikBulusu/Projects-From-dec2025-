```python
sales = [100, 200, 150, 300, 120]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

plt.bar(months,sales) ## remember everyone x axis is horizontal , y axis is vertical 
plt.xlabel('monthly')
plt.ylabel('sales')
plt.title("monthly_sales")
plt.show()
```

```python
# question is to plot a scatter plot and add a grid into it 

sales = [100, 200, 150, 300, 120]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

plt.scatter(months,sales) ## remember everyone x axis is horizontal , y axis is vertical 
plt.grid(True) # alright remmeber that grids should be true not numbers
plt.xlabel('monthly')
plt.ylabel('sales')
plt.title("monthly_sales")
plt.show()
```

```python
ages = [22, 25, 27, 23, 35, 40, 45, 28, 30, 22, 25, 38]

plt.hist(ages,bins=5)
plt.title("monthly_sales")
plt.xlabel("months")
plt.ylabel('sales')
plt.grid(True)
plt.show()
```

```python
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales_2023 = [100, 200, 150, 300, 120]
sales_2024 = [150, 180, 200, 280, 160]


x=np.arange(len(months))
width = 0.4
plt.bar(x-width/2 , sales_2023,months ,width, label = "2023")
plt.bar(x+width/2 , sales_2024,months ,width , label = '2024')
plt.xlabel("yearly_sales")
plt.ylabel("month_wise")
plt.show()
```

### The above is the reason why seaborn was made , to handle such groups properly 

```python
import pandas as pd 
import seaborn as sns
df = pd.DataFrame({
    'months': ['Jan','Feb','Mar','Apr','May'],
    'sales': [100, 200, 150, 300, 120]
})

sns.barplot(data= df , x='months' , y='sales' , color = 'green') # and order doesnt exist , it all depends on how we put the data and key diff here is seaborn directly speaks with your data frames than columns like mlt
sns.lineplot(data=df, x='months' , y='sales' ,color = 'lightgreen')
plt.xlabel('months')
plt.ylabel('sales')
plt.title('Summary')
plt.show()

```
