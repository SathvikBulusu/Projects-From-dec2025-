```python
df = pd.DataFrame({
    'sales': [100, 200, 150, 300, 120]
})

df['sales_doubled'] = df['sales'].apply(lambda x: x*2)

df['sales_category'] = df['sales'].apply(lambda x: 'high' if x>150  else "low")

df['summary'] = df['sales'].apply(lambda  x : 'sale:' + str(x))
print(df)
```
